## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Research Question

Which content pages should be reviewed first for refresh or improvement based on observable search and engagement signals?

Decision

Determine which pages should receive review priority.

Action

A content manager reviews the highest-ranked pages and decides whether to:

Refresh outdated content
Expand thin content
Improve titles and metadata
Improve user engagement
Monitor instead of editing
Unit of Analysis

One content page (one row represents one page).

Output

A ranked list of pages containing:

Refresh score
Suggested action
Reason codes
Confidence level
Cost of a Wrong Recommendation

If the model ranks the wrong pages too highly:

valuable editor time is wasted;
important declining pages may be missed;
opportunities for improving traffic could be delayed.

Because of this, the project is intended to support human decisions, not replace them.

Why Machine Learning Helps

There are many measurable signals that influence whether a page deserves attention. Machine learning can combine these signals more effectively than a small set of fixed rules and produce a better prioritised review queue.

The model ranks pages according to observable evidence available in the dataset. It cannot prove that editing a page causes future improvement because this dataset is observational rather than experimental.

Any recommendations produced by the model should be reviewed by a human before action is taken.
